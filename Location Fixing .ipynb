```python
!pip install reverse_geocoder pandas openpyxl pycountry

import pandas as pd
import reverse_geocoder as rg
import pycountry

# === 1. اقرأ الملف ===

df = pd.read_csv("geometry.csv", sep=';')

# === 2. دالة ترجمة كود الدولة لاسم كامل ===
def get_country_name(cc):
    try:
        return pycountry.countries.get(alpha_2=cc).name
    except:
        return cc   # لو مش لاقي الكود يرجعه زي ما هو

# === 3. دالة تجيب البلد والولاية ===
def get_location(lat, lon):
    try:
        results = rg.search((lat, lon))
        cc = results[0]['cc']              # كود الدولة (مثلاً EG, US)
        state = results[0]['admin1']       # اسم الولاية/المحافظة
        country = get_country_name(cc)     # ترجمة الكود لاسم كامل
        return pd.Series([country, state])
    except:
        return pd.Series([None, None])

# === 4. طبّق على الأعمدة (لاحظ L capital) ===
df[['country', 'state']] = df.apply(lambda row: get_location(row['latitude'], row['longitude']), axis=1)

# === 5. احفظ النتيجة ===
output_path = "/content/with_country_state.xlsx"
df.to_excel(output_path, index=False)

print("✅ الملف اتعمل:", output_path)
```
