```python
import pandas as pd

# لو الفاصل فاصلة
nstb = pd.read_csv("NSTB MODEL.csv", sep=';')
faa = pd.read_csv("FAA MODEL.csv", sep=';')

# لو الفاصل tab
# nstb = pd.read_csv("NSTB MODEL.csv", sep='\t')
# faa = pd.read_csv("FAA MODEL.csv", sep='\t')
```

```python
from rapidfuzz import process, fuzz
import pandas as pd

nstb_models = nstb['Model'].astype(str)
faa_models = faa['Model'].astype(str).tolist()

matches = []
for event_id, model in zip(nstb['Event_Id'], nstb_models):  # غيرت ID → Event_Id
    match, score, _ = process.extractOne(model, faa_models, scorer=fuzz.token_sort_ratio)
    matches.append((event_id, model, match, score))

result = pd.DataFrame(matches, columns=['Event_Id', 'NSTB_Model', 'Closest_FAA_Model', 'Similarity'])
result.to_csv('matched_models_with_event_id.csv', index=False)

print("✅ Done! الملف اتخزن باسم matched_models_with_event_id.csv")
```
